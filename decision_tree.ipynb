```python

import pandas as pd
import joblib
from pathlib import Path
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv("drone_communication_dataset.csv")
df.columns = df.columns.str.strip()
```

```python
features = [
    "altitude",
    "speed_trajectory",
    "signal_strength",
    "round_trip_time",
    "network_traffic_volume"
]

attack_labels = [
    "label_spoofing", "label_mitm", "label_ddos",
    "label_gps_spoofing", "label_malware",
    "label_jamming", "label_protocol_exploit"
]
```

data cleaning


```python
df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")

for col in features + ["label_normal"] + attack_labels:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=features + ["timestamp", "label_normal"])
df = df.drop_duplicates()
df = df.sort_values("timestamp")
df = df.drop_duplicates(subset=["timestamp"], keep="first")
```

```python
df["is_anomaly"] = 1 - df["label_normal"].astype(int)   #define new column (whether given data is anomalous or not)
```

```python
available_labels = [c for c in attack_labels if c in df.columns]
if available_labels:
    attack_present = df[available_labels].fillna(0).eq(1).any(axis=1)
    conflict = (
        ((df["label_normal"] == 1) & attack_present)
        | ((df["label_normal"] == 0) & (~attack_present))
    )
    print("Label conflicts:", int(conflict.sum()))
```

```python
df = df.reset_index(drop=True)
split = int(0.8 * len(df))

train = df.iloc[:split]
test = df.iloc[split:]

X_train = train[features].copy()
X_test = test[features].copy()
y_train = train["is_anomaly"]
y_test = test["is_anomaly"]


medians = X_train.median()
X_train = X_train.fillna(medians)
X_test = X_test.fillna(medians)
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    min_samples_leaf=20,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, pred), 4))
print("Classification report:")
print(classification_report(
    y_test,
    pred,
    labels=[0, 1],
    target_names=["Normal", "Anomaly"],
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_test, pred, labels=[0, 1]))
```

```python
print("\n=== DECISION TREE RULES ===")
print(export_text(model, feature_names=features, decimals=3))


out = Path("droneguard_tree_outputs")
out.mkdir(exist_ok=True)

joblib.dump({
    "model": model,
    "features": features,
    "medians": medians.to_dict()
}, out / "droneguard_decision_tree.joblib")

results = test[["timestamp"] + features].copy()
results["actual_anomaly"] = y_test.to_numpy()
results["predicted_anomaly"] = pred
results["anomaly_probability"] = model.predict_proba(X_test)[:, 1]
results.to_csv(out / "test_predictions.csv", index=False)

df.to_csv(out / "cleaned_dataset.csv", index=False)

print("\nSaved files in droneguard_tree_outputs/")
```
